```python
import pandas as pd , matplotlib.pyplot as plt, seaborn as sb,numpy as np
from sklearn.linear_model import LogisticRegression, LassoCV, ElasticNetCV
from sklearn.model_selection import train_test_split,cross_val_score, KFold,GridSearchCV
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
from sklearn.preprocessing import StandardScaler 
from sklearn.neighbors import KNeighborsClassifier as KNN
from sklearn.ensemble import RandomForestClassifier
```

```python
Data=df.copy()
Data.drop(['fractal dimension error','mean fractal dimension','smoothness error','symmetry error','texture error'],axis=1,inplace=True)
Data.head()
```

```python
X=Data.iloc[ :, :-1]
Y=Data.iloc[: , -1]
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, random_state=33)
```

```python
model=LogisticRegression(max_iter=20000)
model.fit(X_train,Y_train)
Y_pred=model.predict(X_test)
accuracy_updated = accuracy_score(Y_test, Y_pred)
print(classification_report(y_true=Y_test,y_pred=Y_pred))
```

```python
ConfusionMatrix=confusion_matrix(Y_test, Y_pred)
plt.figure()
sb.heatmap(ConfusionMatrix,annot=True,cmap='Greys')
plt.xlabel("Predicted value")
plt.ylabel("Actual value")
plt.title("Confusion matrix")
plt.show()
```

```python
confusion_matrix(Y_test, Y_pred)
```

Model is too acurate which could possibly be a case of over-fitting<br>
Too check whether the model is over-ftted or not I will use k-Fold

```python
model=LogisticRegression(max_iter=20000)

kf=KFold(n_splits=5, shuffle=True, random_state=42)
score=cross_val_score(model,X_train,Y_train,cv=kf,scoring='accuracy')
print(f'Cross-Validation Accuracy Scores: {score}')
print(f'Mean CV Accuracy: {score.mean()}')

model.fit(X_train,Y_train)
Y_pred=model.predict(X_test)
print(classification_report(y_true=Y_test,y_pred=Y_pred))
```

Since the cross-calidation accuracy scores are consistantly high this indicates that the model is stable and not over-fitting the data 

```python
def Regularization(reg):
    reg.fit(X_train,Y_train)
    print("Best alpha using built-in {}: %f".format(str(reg).split('.')[-1]) % reg.alpha_)
    print("Best score using built-in {}: %f".format(str(reg).split('.')[-1]) %reg.score(X_train,Y_train))
    coef = pd.Series(reg.coef_, index = X_train.columns)
    print("{} picked ".format(str(reg).split('.')[-1]) + str(sum(coef != 0)) + " variables and eliminated the other " +  str(sum(coef == 0)) + " variables")
    coef = pd.Series(reg.coef_ , index=X_train.columns)
    imp_coef = coef.sort_values()
    print(imp_coef)
Regularization(ElasticNetCV())
Regularization(LassoCV())
```
